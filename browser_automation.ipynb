## 2. Seitenanalyse-Funktionen

Simuliert eine Browser-Use-Seitenanalyse (in der Live-Version mit echtem Browser).

```python
def generate_page_analysis(url: str) -> dict:
    """Simuliert eine Browser-Use Seitenanalyse."""
    domain = url.replace('https://', '').replace('http://', '').split('/')[0]
    
    return {
        'url': url,
        'domain': domain,
        'title': f'Seite: {domain}',
        'status_code': 200,
        'load_time_ms': hash(url) % 2000 + 200,
        'content_length_kb': hash(url) % 5000 + 50,
        'has_https': url.startswith('https'),
        'analyzed_at': datetime.now().isoformat(),
    }


def generate_elements(url: str) -> List[dict]:
    """Simuliert extrahierte Seitenelemente."""
    domain = url.replace('https://', '').replace('http://', '').split('/')[0]
    
    elements = [
        {'Typ': 'Überschrift (H1)', 'Inhalt': f'Willkommen auf {domain}', 'Selektor': 'h1'},
        {'Typ': 'Absatz', 'Inhalt': f'Dies ist eine Beispiel-Seite für {domain}. Hier finden Sie alle wichtigen Informationen.', 'Selektor': 'p.main'},
        {'Typ': 'Link', 'Inhalt': 'https://example.com/mehr', 'Selektor': 'a.nav-link'},
        {'Typ': 'Bild', 'Inhalt': f'https://{domain}/logo.png', 'Selektor': 'img.logo'},
        {'Typ': 'Button', 'Inhalt': 'Jetzt kaufen', 'Selektor': 'button.cta'},
        {'Typ': 'Eingabefeld', 'Inhalt': 'placeholder: Suche...', 'Selektor': 'input.search'},
        {'Typ': 'Liste', 'Inhalt': '3 Einträge gefunden', 'Selektor': 'ul.results'},
        {'Typ': 'Formular', 'Inhalt': 'Kontaktformular', 'Selektor': 'form#contact'},
    ]
    
    return elements


def generate_links(url: str, count: int = 8) -> List[dict]:
    """Simuliert extrahierte Links."""
    domain = url.replace('https://', '').replace('http://', '').split('/')[0]
    links = []
    for i in range(count):
        links.append({
            'Text': f'Link {i+1}: Seite {i+1}',
            'URL': f'https://{domain}/page{i+1}',
            'Typ': 'intern' if i % 3 != 0 else 'extern',
        })
    return links

# Test mit Beispiel-URLs
TEST_URLS = [
    'https://example.com',
    'https://www.github.com',
    'https://www.wikipedia.org',
]

for url in TEST_URLS:
    analysis = generate_page_analysis(url)
    print(f'✅ {analysis["domain"]}: Status {analysis["status_code"]}, Ladezeit {analysis["load_time_ms"]}ms')
```

## 3. URL-Analyse durchführen

Detaillierte Analyse einer einzelnen URL mit allen Metadaten.

```python
# Konfiguration
TARGET_URL = 'https://example.com'
VIEWPORT_WIDTH = 1920
VIEWPORT_HEIGHT = 1080
WAIT_TIME = 3  # Sekunden

# Analyse durchführen
analysis = generate_page_analysis(TARGET_URL)
elements = generate_elements(TARGET_URL)
links = generate_links(TARGET_URL)

print(f'🌐 Seitenanalyse: {TARGET_URL}')
print(f'{"─" * 60}')

# Übersicht
print(f'\n📋 Übersicht:')
print(f'  Domain:        {analysis["domain"]}')
print(f'  Titel:         {analysis["title"]}')
print(f'  Status:        {analysis["status_code"]} OK')
print(f'  Ladezeit:      {analysis["load_time_ms"]} ms')
print(f'  Größe:         {analysis["content_length_kb"]} KB')
print(f'  HTTPS:         {"✅" if analysis["has_https"] else "❌"}')
print(f'  Viewport:      {VIEWPORT_WIDTH}×{VIEWPORT_HEIGHT}')
print(f'  Wartezeit:     {WAIT_TIME}s')

# Browser-Info
print(f'\n🌐 Browser-Informationen:')
browser_info = {
    'browser': 'Chromium (headless)',
    'version': '120.0.6099.109',
    'viewport': f'{VIEWPORT_WIDTH}x{VIEWPORT_HEIGHT}',
    'user_agent': 'Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36',
    'javascript_enabled': True,
    'cookies_enabled': True,
}
for key, value in browser_info.items():
    print(f'  {key}: {value}')
```

## 4. Element-Extraktion

Strukturierte Extraktion von DOM-Elementen: Überschriften, Links, Bilder, Buttons, Formulare.

```python
print(f'🔍 Extrahierte Elemente ({len(elements)}):')
print(f'{"─" * 60}')

df_elements = pd.DataFrame(elements)
df_elements

# Element-Statistik
print(f'\n📊 Element-Statistik:')
type_counts = {}
for e in elements:
    main_type = e['Typ'].split(' (')[0]
    type_counts[main_type] = type_counts.get(main_type, 0) + 1

df_stats = pd.DataFrame({
    'Element-Typ': list(type_counts.keys()),
    'Anzahl': list(type_counts.values()),
})
df_stats
```

## 5. Element-Statistik visualisieren

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Balkendiagramm: Element-Typen
types = list(type_counts.keys())
counts = list(type_counts.values())
colors = ['#3498db', '#2ecc71', '#e74c3c', '#f39c12', '#9b59b6', '#1abc9c', '#e67e22', '#34495e']
ax1.barh(types, counts, color=colors[:len(types)])
ax1.set_xlabel('Anzahl')
ax1.set_title(f'DOM-Elemente auf {analysis["domain"]}', fontsize=13, fontweight='bold')
for i, v in enumerate(counts):
    ax1.text(v + 0.1, i, str(v), va='center', fontweight='bold')

# Pie-Chart: Interne vs. Externe Links
link_types = {}
for link in links:
    link_types[link['Typ']] = link_types.get(link['Typ'], 0) + 1

ax2.pie(list(link_types.values()), labels=list(link_types.keys()),
        autopct='%1.1f%%', colors=['#3498db', '#e74c3c'], startangle=90)
ax2.set_title('Link-Verteilung (Intern vs. Extern)', fontsize=13, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 6. Screenshot-Informationen

In der Live-Version wird hier ein echter Screenshot mit Browser-Use + Playwright erstellt.

```python
print(f'📸 Screenshot: {analysis["domain"]}')
print(f'{"─" * 60}')

screenshot_info = {
    'format': 'PNG',
    'full_page': False,
    'viewport': f'{VIEWPORT_WIDTH}x{VIEWPORT_HEIGHT}',
    'timestamp': analysis['analyzed_at'],
    'file_size_estimate': f'{analysis["content_length_kb"] * 3} KB',
    'status': '📸 Screenshot wird in der Live-Version mit Browser-Use + Playwright erstellt.',
}

for key, value in screenshot_info.items():
    print(f'  {key}: {value}')

# Screenshot-Platzhalter als farbige Box
print(f'\n{"─" * 60}')
print(f'│  📸 {analysis["domain"]:^50} │')
print(f'│  Viewport: {VIEWPORT_WIDTH}×{VIEWPORT_HEIGHT:^38} │')
print(f'│  Ladezeit: {analysis["load_time_ms"]}ms{" ":^38} │')
print(f'{"─" * 60}')
```

## 7. Multi-URL-Vergleich

Vergleich mehrerer URLs hinsichtlich Ladezeit, Größe und HTTPS-Status.

```python
# Mehrere URLs analysieren
comparison_urls = [
    'https://example.com',
    'https://www.github.com',
    'https://www.wikipedia.org',
    'https://www.python.org',
    'https://stackoverflow.com',
]

results = []
for url in comparison_urls:
    a = generate_page_analysis(url)
    results.append(a)

df_compare = pd.DataFrame([{
    'Domain': r['domain'],
    'Status': r['status_code'],
    'Ladezeit (ms)': r['load_time_ms'],
    'Größe (KB)': r['content_length_kb'],
    'HTTPS': '✅' if r['has_https'] else '❌',
} for r in results])

print('📊 Multi-URL-Vergleich:')
df_compare

# Visualisierung
fig, ax = plt.subplots(figsize=(12, 5))

x = range(len(comparison_urls))
width = 0.35

bars1 = ax.bar([i - width/2 for i in x], [r['load_time_ms'] for r in results],
               width, label='Ladezeit (ms)', color='#3498db')
ax.set_ylabel('Ladezeit (ms)', color='#3498db', fontsize=12)
ax.tick_params(axis='y', labelcolor='#3498db')

ax2 = ax.twinx()
bars2 = ax2.bar([i + width/2 for i in x], [r['content_length_kb'] for r in results],
                width, label='Größe (KB)', color='#e74c3c')
ax2.set_ylabel('Größe (KB)', color='#e74c3c', fontsize=12)
ax2.tick_params(axis='y', labelcolor='#e74c3c')

ax.set_xticks(x)
ax.set_xticklabels([r['domain'] for r in results], rotation=30, ha='right')
ax.set_title('Ladezeit & Seitengröße im Vergleich', fontsize=14, fontweight='bold')

# Legende
lines1, labels1 = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.tight_layout()
plt.show()
```

## 8. Analyse-Verlauf

Tracking aller durchgeführten Analysen mit Zeitstempel.

```python
# Verlauf aus allen Analysen erstellen
all_analyses = [generate_page_analysis(url) for url in comparison_urls]

df_history = pd.DataFrame([{
    'Zeitpunkt': a['analyzed_at'],
    'URL': a['url'],
    'Domain': a['domain'],
    'Status': a['status_code'],
    'Ladezeit (ms)': a['load_time_ms'],
    'Größe (KB)': a['content_length_kb'],
    'HTTPS': '✅' if a['has_https'] else '❌',
} for a in all_analyses])

print(f'📜 Analyse-Verlauf ({len(df_history)} Einträge):')
df_history

# Zusammenfassung
print(f'\n📊 Zusammenfassung:')
print(f'  Durchschnittliche Ladezeit: {df_history["Ladezeit (ms)"].mean():.0f} ms')
print(f'  Durchschnittliche Größe:    {df_history["Größe (KB)"].mean():.0f} KB')
print(f'  HTTPS-Quote:                {(df_history["HTTPS"] == "✅").sum()}/{len(df_history)}')
```
